```python
import os

import pandas as pd
import numpy as np
from glob import glob
```

```python
cache_dir = "./wandb_cache"
```

```python
csvs = glob(f"{cache_dir}/*.csv")
```

```python
#  concat 
df = pd.concat([pd.read_csv(csv) for csv in csvs])
df = df[df['target_character'] == 2]
df
```

```python
# Filter data where Evaluation/llm_iteration == 1 and Evaluation/acc_imp_perc == 0.0
iteration_1 = df[(df['Evaluation/llm_iteration'] == 1) & (df['Evaluation/acc_imp_perc'] < 0.2)]

# Filter data where Evaluation/llm_iteration == 6 and Evaluation/acc_imp_perc == 1.0
iteration_6 = df[(df['Evaluation/llm_iteration'] > 5) & (df['Evaluation/acc_imp_perc'] >= 0.8)]



# Find run_ids that meet both conditions
filtered_run_ids = set(iteration_1['run_id']).intersection(set(iteration_6['run_id']))

# Convert the result to a DataFrame for easier handling
df_filtered = df[df['run_id'].isin(filtered_run_ids)]
df_filtered
```

```python
# download image from wandb
import wandb
import os
PROJECT_NAME = "scenario"
ENTITY_NAME = "inchangbaek4907"
```

```python
cherry_dir = "cherry"
os.makedirs(cherry_dir, exist_ok=True)

for idx, row in df_filtered.iterrows():

    api = wandb.Api()
    run = api.run(f"{ENTITY_NAME}/{PROJECT_NAME}/{row['run_id']}")
    
    base_url = run.files()[0].client.app_url
    files = run.files()
    for file in files:
        name = file.name
        url = file.directUrl
        
    
        # if not name includes "rollout" and "png", skip it
        if "rollout" not in name or "png" not in name:
            continue
        
        iteration = name.split("/")[2]
        filename = name.split("/")[-1]
        
        origin_path = os.path.join(".temp", 'media', 'images', iteration, 'rollout', filename)
        target_path = os.path.join(cherry_dir, f'chr_{row["target_character"]}', row["run_id"], iteration)
        

        file.download(exist_ok=True, root=".temp")

        # copy file
        os.makedirs(target_path, exist_ok=True)
        os.system(f"cp {origin_path} {target_path}")
        print(f"cp {origin_path} {target_path}")
        
```
