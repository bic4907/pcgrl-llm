```python
# min-max normalization with 'score' column for target_character-wise
def min_max_normalize(df):
    df = df.copy()
    def normalize_group(sub_df):
        min_val = sub_df['score'].min()
        max_val = sub_df['score'].max()
        if max_val > min_val:
            sub_df['score'] = (sub_df['score'] - min_val) / (max_val - min_val)
        else:
            sub_df['score'] = 0.0  # 모든 값이 동일한 경우
        return sub_df
    return df.groupby('target_character').apply(normalize_group).reset_index(drop=True)

normalized = min_max_normalize(df)
normalized
```

```python
# average the target_character
avg_df = normalized.groupby(['gpt_model', 'pe', 'n_self_alignment', 'Evaluation/llm_iteration', 'seed']).agg({'score': 'mean'}).reset_index()
avg_df['target_character'] = 'mean'
avg_df
```

```python
import pandas as pd
import numpy as np
try:
    from scipy.stats import ttest_ind
    _HAS_SCIPY = True
except Exception:
    _HAS_SCIPY = False

def _welch_ttest_p(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)
    a = a[~np.isnan(a)]
    b = b[~np.isnan(b)]
    if len(a) < 2 or len(b) < 2:
        return np.nan
    if _HAS_SCIPY:
        return ttest_ind(a, b, equal_var=False, nan_policy='omit').pvalue
    # fallback (정규 근사)
    ma, mb = a.mean(), b.mean()
    va, vb = a.var(ddof=1), b.var(ddof=1)
    na, nb = len(a), len(b)
    se2 = va/na + vb/nb
    if se2 == 0:
        return np.nan
    t = (ma - mb) / np.sqrt(se2)
    from math import erf, sqrt
    p = 2 * (1 - 0.5*(1 + erf(abs(t)/sqrt(2))))
    return p
```

```python
def make_table(df: pd.DataFrame, ref_label: str = "PCGRLLM") -> pd.DataFrame:
    def agg_stats(sub_df, label):
        grouped = (
            sub_df.groupby(['gpt_model', 'pe'])
            .agg({'score': ['mean', 'std']})
            .reset_index()
        )
        grouped.columns = ['gpt_model', 'pe', f'{label} (mean)', f'{label} (std)']
        return grouped

    # 각 mode 정의
    zero_mask  = (df['n_self_alignment'] == 0) & (df['Evaluation/llm_iteration'] == 1)
    sa_mask    = (df['n_self_alignment'] == 5) & (df['Evaluation/llm_iteration'] == 1)
    fb_mask    = (df['n_self_alignment'] == 0) & (df['Evaluation/llm_iteration'] == 6)
    safb_mask  = (df['n_self_alignment'] == 5) & (df['Evaluation/llm_iteration'] == 6)

    base_df  = agg_stats(df[zero_mask], "Zero-shot")
    sa_df    = agg_stats(df[sa_mask],   "+Self-Alignment")
    fb_df    = agg_stats(df[fb_mask],   "+Feedback")
    safb_df  = agg_stats(df[safb_mask], "PCGRLLM")

    pk = ['gpt_model', 'pe']
    merged = (
        base_df.merge(sa_df, on=pk, how='outer')
               .merge(fb_df, on=pk, how='outer')
               .merge(safb_df, on=pk, how='outer')
    )

    # --- p-value 계산 ---
    p_rows = []
    for (g, p), _ in df.groupby(['gpt_model', 'pe']):
        sub = df[(df['gpt_model'] == g) & (df['pe'] == p)]

        zs_cur = sub[zero_mask]['score'].to_numpy()
        sa_cur = sub[sa_mask]['score'].to_numpy()
        ref    = sub[safb_mask]['score'].to_numpy()  # PCGRLLM 기준

        p_rows.append({
            'gpt_model': g,
            'pe': p,
            f'p: Zero-shot vs {ref_label}': _welch_ttest_p(zs_cur, ref),
            f'p: +Self-Alignment vs {ref_label}': _welch_ttest_p(sa_cur, ref),
        })

    pvals_df = pd.DataFrame(p_rows)
    merged = merged.merge(pvals_df, on=pk, how='left')
    return merged

```

```python
def make_perf_table_by_targets(df, targets=(1, 2, 5, 6, 'mean')):
    tables = []
    for t in targets:
        
        sub = df[df['target_character'] == t]
        
        if sub.empty:
            continue
        tbl = make_table(sub)           # 이미 mean/std 포함 버전
        tbl.insert(0, 'target_character', t)
        tables.append(tbl)
    if not tables:
        return pd.DataFrame()
    return pd.concat(tables, ignore_index=True)
```

```python
_df = df[df['gpt_model'] == 'gpt-4o']
perf_table_gpt = make_perf_table_by_targets(_df)
perf_table_gpt
```

```python
import numpy as np
import pandas as pd

def df_to_latex_table(df: pd.DataFrame) -> str:
    """
    평균 ± 표준편차 형태로 출력,
    Zero-shot 포함 모든 조건 중 최고 mean은 bold,
    p < 0.05이면 마크 부여:
      - +Self-Alignment vs PCGRLLM : * (PCGRLLM 칸에 표시)
      - Zero-shot vs PCGRLLM       : † (PCGRLLM 칸에 표시)
    Target 값이 바뀔 때마다 \midrule 삽입
    셀 색상은 gray!n 사용, 셀 높이는 strut으로 고정
    """
    # 컬럼명 축약
    rename_map = {
        "p: Zero-shot vs PCGRLLM": "p(ZS)",
        "p: +Self-Alignment vs PCGRLLM": "p(SA)",
        "Zero-shot (mean)": "Zero-shot (ZS) (mean)",
        "Zero-shot (std)": "Zero-shot (ZS) (std)",
        "+Self-Alignment (mean)": "+Self-Alignment (SA) (mean)",
        "+Self-Alignment (std)": "+Self-Alignment (SA) (std)",
    }
    df = df.rename(columns={k: v for k, v in rename_map.items() if k in df.columns})

    mean_std_cols = [c for c in df.columns if "(mean)" in c]
    p_val_cols = [c for c in df.columns if c.startswith("p(")]
    cols = mean_std_cols + p_val_cols
    header = ["Target", "PE"] + [c.replace(" (mean)", "").replace(" (std)", "") for c in cols]

    # 정렬: Target 오름차순, PE = CoT → ToT → GoT
    pe_map = {"cot": "CoT", "tot": "ToT", "got": "GoT"}
    pe_order = {"CoT": 0, "ToT": 1, "GoT": 2}
    df["pe"] = df["pe"].map(pe_map)
    df = df.sort_values(
        by=["target_character", "pe"],
        key=lambda col: col.map(pe_order) if col.name == "pe" else col
    )

    # --- Global min/max for coloring ---
    all_means = df[mean_std_cols].to_numpy().flatten()
    vmin, vmax = np.nanmin(all_means), np.nanmax(all_means)

    def val_to_gray(val):
        """값을 vmin~vmax 구간에서 회색 강도로 매핑 (gray!5 ~ gray!30)"""
        if pd.isna(val):
            return "white"
        norm = (val - vmin) / (vmax - vmin + 1e-8)
        level = int(5 + norm * 25)  # 5~30 사이
        return f"gray!{level}"

    latex_rows = []
    prev_target = None
    for _, row in df.iterrows():
        curr_target = str(row.get("target_character", ""))

        if prev_target is not None and curr_target != prev_target:
            latex_rows.append("\\midrule")
        prev_target = curr_target

        row_entries = [curr_target, str(row["pe"])]

        # 최고 mean bold
        means = [row[c] for c in mean_std_cols]
        best_idx = int(np.nanargmax(means)) if len(means) > 0 else None
        best_label = mean_std_cols[best_idx].replace(" (mean)", "") if best_idx is not None else None

        for c in mean_std_cols:
            label = c.replace(" (mean)", "")
            mean = row[c]
            std = row[c.replace("(mean)", "(std)")]
            color = val_to_gray(mean)

            # strut으로 높이 고정
            base = f"\\cellcolor{{{color}}}\\rule{{0pt}}{{2.2ex}}\\color{{black}} {mean:.3f}$\\pm${{\\scriptsize {std:.3f}}}"

            # Bold
            if label == best_label:
                base = f"\\textbf{{{base}}}"

            # Significance mark → PCGRLLM 칸에만
            if label == "PCGRLLM":
                if "p(ZS)" in row and pd.notna(row["p(ZS)"]):
                    if row["p(ZS)"] < 0.05:
                        base += "$^{\\dagger}$"
                if "p(SA)" in row and pd.notna(row["p(SA)"]):
                    if row["p(SA)"] < 0.05:
                        base += "$^{*}$"

            row_entries.append(base)

        # p-value 셀 (숫자만, 중앙 정렬, 높이 통일)
        for c in p_val_cols:
            val = row[c]
            cell = "" if pd.isna(val) else f"\\rule{{0pt}}{{2.2ex}}{val:.3f}"
            row_entries.append(cell)

        latex_rows.append(" & ".join(row_entries) + " \\\\")

    # colspec: p-value 중앙정렬
    col_spec = "p{1.5cm}p{1.5cm}p{2.6cm}p{2.6cm}p{2.6cm}p{2.6cm}|cc"

    table = ""
    table += "\\begin{table*}\n"
    table += "\\centering\n"
    table += "\\caption{Performance by PE and Target. Significance marks: "
    table += "$^{*}$ for +Self-Alignment vs PCGRLLM, $^{\\dagger}$ for Zero-shot vs PCGRLLM.}\n"
    table += "\\label{tab:performance_pcgrllm}\n"
    table += f"\\begin{{tabular}}{{{col_spec}}}\n"
    table += "\\toprule\n"
    table += " & ".join(header) + " \\\\\n\\midrule\n"
    table += "\n".join(latex_rows) + "\n"
    table += "\\bottomrule\n\\end{tabular}\n"
    table += "\\end{table*}\n"
    return table
```
